### 3. Dataset Description

The dataset contains retail transaction records from the Global Superstore business. Each row represents a recorded transaction line containing information about an order and its associated product.

The dataset includes the following categories of information:

- **Order Information:** Order ID, order date, shipping date, and order priority.
- **Customer Information:** Customer ID, customer name, and customer segment.
- **Product Information:** Product ID, product name, category, and sub-category.
- **Sales Information:** Sales, quantity, discount, and profit.
- **Geographical Information:** City, state, country, market, and region.
- **Shipping Information:** Shipping mode and shipping cost.
