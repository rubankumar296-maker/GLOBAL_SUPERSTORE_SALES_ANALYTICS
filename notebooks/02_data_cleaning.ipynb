## Data Cleaning and Preprocessing

### 1. Introduction

Data cleaning is an essential step in data analytics because the quality of the data directly affects the accuracy of the analysis and business insights.

In this phase, the original Global Superstore dataset will be inspected, cleaned, and transformed into a reliable dataset for Exploratory Data Analysis (EDA), SQL business analysis, and Power BI dashboard development.

The raw dataset contains information about sales transactions, customers, products, order dates, shipping details, regions, markets, discounts, and profits.

### 2. Objectives

The main objectives of this phase are:

- Inspect the dataset for data quality issues.
- Identify and handle missing or invalid values where necessary.
- Check for duplicate records and investigate potential duplicate identifiers.
- Standardize column names and text values where appropriate.
- Convert columns to their correct data types.
- Validate numerical values, order dates, and shipping dates.
- Investigate unusual values without automatically removing valid business records.
- Create useful date-based features for future analysis.
- Save the cleaned dataset for subsequent project phases.

### 3. Tools and Libraries

The following tools and libraries will be used:

- **Python:** Data processing and validation.
- **Pandas:** Data manipulation and cleaning.
- **NumPy:** Numerical operations and missing-value handling.
- **Matplotlib and Seaborn:** Visual inspection when required.

### 4. Data Cleaning Workflow

The data cleaning process will follow these steps:

1. Load the original dataset.
2. Create a working copy to preserve the raw data.
3. Inspect columns, data types, missing values, and duplicates.
4. Clean text fields and standardize column values where necessary.
5. Validate and convert date and numerical columns.
6. Investigate unusual values and data consistency issues.
7. Create useful date-related features.
8. Perform final data quality checks.
9. Export the cleaned dataset to the processed data folder.

### 5. Expected Outcome

The expected outcome is a validated and consistently formatted dataset saved as:

`data/processed/superstore_cleaned.csv`

This cleaned dataset will be used in **Exploratory Data Analysis (EDA)**, followed by SQL business analysis and Power BI dashboard development.

**Important:** The original raw dataset will remain unchanged. Cleaning decisions will be based on actual data quality findings rather than assumptions.

```python
import pandas as pd
import numpy as np
from pathlib import Path

input_path = Path("../data/raw/Superstore.csv")

df = pd.read_csv(input_path)

print("Original dataset shape:", df.shape)
df.head()
```

```python
# Create a working copy

df_clean = df.copy()

print("Working copy created.")
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))
```

```python
# Investigate Unnamed: 0

if "Unnamed: 0" in df_clean.columns:
    print("First values:", df_clean["Unnamed: 0"].head().tolist())
    print("Unique values:", df_clean["Unnamed: 0"].nunique())
    print(
        "Matches sequential index:",
        df_clean["Unnamed: 0"].equals(
            pd.Series(range(len(df_clean)), index=df_clean.index)
        )
    )

if (
    "Unnamed: 0" in df_clean.columns
    and df_clean["Unnamed: 0"].equals(
        pd.Series(range(len(df_clean)), index=df_clean.index)
    )
):
    df_clean = df_clean.drop(columns=["Unnamed: 0"])
    print("Removed the redundant index column.")
else:
    print("Kept the column for further investigation.")
```

```python
# Clean text columns

text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for column in text_columns:
    df_clean[column] = df_clean[column].apply(
        lambda value: value.strip()
        if isinstance(value, str)
        else value
    )

print("Text fields cleaned.")
```

```python
# Convert date columns

date_columns = ["Order_Date", "Ship_Date"]

for column in date_columns:
    df_clean[column] = pd.to_datetime(
        df_clean[column],
        errors="coerce"
    )

print(df_clean[date_columns].dtypes)

for column in date_columns:
    print(
        column,
        "Invalid dates:",
        df_clean[column].isna().sum()
    )
```

```python
# Validate numeric columns

numeric_columns = [
    "Sales",
    "Profit",
    "Quantity",
    "Discount",
    "Shipping_Cost",
    "Record_Count",
    "Row_ID",
    "Year",
    "Weeknum"
]

print(df_clean[numeric_columns].dtypes)

# Check for infinite values

for column in numeric_columns:
    values = pd.to_numeric(df_clean[column], errors="coerce")
    print(
        column,
        "| Invalid:", values.isna().sum(),
        "| Infinite:", np.isinf(values).sum()
    )
```

```python
# Check duplicate identifiers and date logic 

print("Duplicate Row_ID values:", df_clean["Row_ID"].duplicated().sum())
print("Unique Order_ID values:", df_clean["Order_ID"].nunique())
print("Unique Customer_ID values:", df_clean["Customer_ID"].nunique())

# check shipping date

invalid_shipping_dates = df_clean[
    df_clean["Ship_Date"] < df_clean["Order_Date"]
]

print("Ship date before order date:", len(invalid_shipping_dates))
```

```python
# Investigate business values

print("Negative-profit records:", (df_clean["Profit"] < 0).sum())
print("Zero-sales records:", (df_clean["Sales"] == 0).sum())
print("Negative-sales records:", (df_clean["Sales"] < 0).sum())
print("Zero-quantity records:", (df_clean["Quantity"] == 0).sum())
print("Negative-quantity records:", (df_clean["Quantity"] < 0).sum())
```

```python
# Create useful date features

df_clean["Order_Year"] = df_clean["Order_Date"].dt.year

df_clean["Order_Month"] = df_clean["Order_Date"].dt.month

df_clean["Month_Name"] = df_clean["Order_Date"].dt.month_name()

df_clean["Year_Month"] = (
    df_clean["Order_Date"].dt.to_period("M").astype(str)
)

df_clean["Shipping_Days"] = (
    df_clean["Ship_Date"] - df_clean["Order_Date"]
).dt.days

print("Date features created.")
df_clean[
    ["Order_Date", "Order_Year", "Month_Name",
     "Year_Month", "Shipping_Days"]
].head()
```

```python
# Final validation

print("Final shape:", df_clean.shape)
print("Missing values:", df_clean.isna().sum().sum())
print("Exact duplicate rows:", df_clean.duplicated().sum())

df_clean.info()
```

```python
# Save the cleaned dataset

output_dir = Path("../data/processed")
output_dir.mkdir(parents=True, exist_ok=True)

output_path = output_dir / "superstore_cleaned.csv"

df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved to:", output_path)
```
